- Passo 5: Comparar os Resultados dos Modelos e Escolher o Vencedor

- Passo 7: Fazer ajustes no Melhor Modelo 

```python
import pandas as pd
import numpy as np
from sklearn.metrics import r2_score, mean_squared_error 
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor
from sklearn.model_selection import train_test_split
```

```python
# Ler a base

base_cod = pd.read_parquet('base_cod.parquet')
```

#### Métricas de avaliação dos modelos
- R2 - variância (quanto MAIOR melhor)
- RSME - erro quadrático médio (quanto MENOR melhor)

```python
def avaliacao_modelo(nome_modelo, y_teste, previsao):
    r2 = r2_score(y_teste, previsao)
    RSME = np.sqrt(mean_squared_error(y_teste, previsao))
    
    return f'Modelo {nome_modelo}:\n R2 {r2}:\n RSME {RSME}'
```

### Modelos de Regressão IA
- Linear Regression
- Random Forest Regressor
- Extra Trees

```python
modelo_rf = RandomForestRegressor()
modelo_lr = LinearRegression()
modelo_et = ExtraTreesRegressor()

modelos = {'RandomForest': modelo_rf,
           'LinearRegression': modelo_lr,
           'ExtraTrees': modelo_et
           }

y = base_cod['price']
x = base_cod.drop('price', axis = 1)
```

### Separa os dados em Treino e Teste

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, random_state=10)  # usa o random_state para gerar sempre os mesmos valores de treino e teste.
```

### Treinar os Modelos

```python
for nome_modelo, modelo in modelos.items():
    #treinar
    modelo.fit(x_train, y_train)
    #testar
    previsao = modelo.predict(x_test)

    print(avaliacao_modelo(nome_modelo, y_test, previsao))
    
```
