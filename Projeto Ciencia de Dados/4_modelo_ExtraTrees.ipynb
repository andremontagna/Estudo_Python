### O Modelo escolhido foi o Extre Trees

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.model_selection import train_test_split
import time
from funcao_teste import avaliacao_modelo
import joblib
```

```python
base_cod = pd.read_parquet('base_cod.parquet')

y = base_cod['price']
x = base_cod.drop('price', axis = 1)
```

```python
modelo_et = ExtraTreesRegressor()
modelo = modelo_et

x_train, x_test, y_train, y_test = train_test_split(x,y, random_state=10)  # usa o random_state para gerar sempre os mesmos valores de treino e teste.
```

```python
def treinar_e_avaliar_et(modelo, x_train, x_test, y_train, y_test):
    inicio = time.time()
    
    modelo.fit(x_train, y_train)
    previsao = modelo.predict(x_test)
    
    importancia_features = pd.DataFrame(modelo.feature_importances_, 
                                        index=x_train.columns, 
                                        columns=['Importancia'])
    importancia_features = importancia_features.sort_values(by='Importancia', ascending=False)
    
    fim = time.time()
    tempo_total = fim - inicio
    
    print(f"Tempo de execução: {tempo_total:.2f} segundos")
    # Chamando sua função de avaliação externa
    print(avaliacao_modelo("Extra Trees", y_test, previsao))
    print("-" * 30)
    
    return importancia_features
```

```python
importancia = treinar_e_avaliar_et(modelo, x_train, x_test, y_train, y_test)

# O gráfico de importância agora pode ser visualizado:
print(importancia.head())

```
