### Função de pesquisa no site Gutenberg

```python
def pesquisa_gutenberg (nome, autor, navegador):
    navegador.get("https://gutenberg.org") #acessa o site

    busca = navegador.find_element("class name", "search-input") #encontrar o elemento na página (botão de busca) 
    busca.send_keys(nome) # insere o item de pesquisa
    busca.send_keys(Keys.ENTER)

    lista_resultados= navegador.find_elements("class name", "booklink") # seleciona todos os resultados da pesquisa
    link_resultado= None
    for resultado in lista_resultados: 
        texto = resultado.text #lista os resultdos pelo nome
        if nome.lower() in texto.lower(): # verifica se o nome foi encontrado
            if all(palavra in texto for palavra in autor.split(" ")): #verifica se o autor está contido no texto, mesmo que nome não completo
                link_resultado = resultado.find_element("class name","link").get_attribute("href")
                
                break
    if link_resultado:
        preco = 0
    else:
        preco = 0
    
    return link_resultado, preco

link, preco = pesquisa_gutenberg(nome, autor, navegador)

print(link)
print(f'R$ {preco: .2f}')
```

```python
def pesquisa_books (nome, categoria, navegdor):
    navegador.get('https://books.toscrape.com/')

    lista_categorias = navegador.find_element('class name', 'nav-list')
    try:
        lista_categorias.find_element('link text', categoria).click()
    except:
        print('Categoria não encontrada')

    lista_resultados= navegador.find_elements("class name", "product_pod") # seleciona todos os resultados da pesquisa
    link_livro = None
    for resultado in lista_resultados: 
        titulo = resultado.find_element('tag name', 'h3').find_element('tag name', 'a').get_attribute('title')
        if nome.lower() in titulo.lower():
            link_livro = resultado.find_element('tag name', 'h3').find_element('tag name', 'a').get_attribute("href")
        break
    
    return (titulo, link_livro)

titulo,link = pesquisa_books(nome, categoria, navegador)

print(titulo)
print(link)
```
